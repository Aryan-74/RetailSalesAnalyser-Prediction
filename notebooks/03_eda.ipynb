# 03. Exploratory Data Analysis (EDA)

Analyze sales trends, seasonality, category distributions, store performance, and promotional elasticity.

```python
import pandas as pd
import config
from src.analytics import (
    get_overall_kpis, plot_daily_sales_trend, plot_weekly_sales_trend,
    plot_monthly_sales_trend, plot_top_families, plot_store_sales,
    plot_promotion_vs_sales, plot_holiday_impact, plot_day_of_week_pattern,
    plot_monthly_seasonality_heatmap, plot_correlation_heatmap
)

df = pd.read_parquet(config.CLEANED_DATA_PATH)
kpis = get_overall_kpis(df)
for k, v in kpis.items():
    print(f'{k}: {v}')
```

## Daily and Weekly Sales Trends

```python
fig_daily = plot_daily_sales_trend(df)
fig_daily.show()
fig_weekly = plot_weekly_sales_trend(df)
fig_weekly.show()
```

## Product Family and Store Breakdown

```python
fig_fams = plot_top_families(df, top_n=10)
fig_fams.show()
fig_stores = plot_store_sales(df)
fig_stores.show()
```

## Promotion and Holiday Impacts

```python
fig_promo = plot_promotion_vs_sales(df)
fig_promo.show()
fig_hol = plot_holiday_impact(df)
fig_hol.show()
```

## Seasonality and Correlations

```python
fig_heat = plot_monthly_seasonality_heatmap(df)
fig_heat.show()
fig_corr = plot_correlation_heatmap(df)
fig_corr.show()
```
